Bar Graph

```python
import matplotlib.pyplot as plt

#BAR CHART
#Step 1: Find total revenue for each city (Capital 'C' in 'City')
city_revenue = df.groupby("City")["Final_Amount"].sum()

#Step 2: Arrange cities from highest to lowest revenue (fixed typo in 'city_revenue')
city_revenue = city_revenue.sort_values(ascending=False)

#Step 3: Print the revenue of each city
print(city_revenue)

#Step 4: Create a bar chart
city_revenue.plot(kind="bar")

#Step 5: Add chart title and y-axis label (fixed string syntax)
plt.title("Total Revenue by City")
plt.ylabel("Total Revenue (Rs)")

#Step 6: Display the chart
plt.show()
```

Line Chart

```python
#Step 1: Convert Order_Date into a date format
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

#Step 2: Count the number of orders each month (joined lines to fix indexing)
monthly_orders = df.groupby(df["Order_Date"].dt.to_period("M"))["Order_ID"].count()

#Step 3: Print the monthly orders
print(monthly_orders)

#Step 4: Create a line chart
monthly_orders.plot(kind="line", marker="o")

#Step 5: Add title and labels
plt.title("Monthly Demand Trend")
plt.ylabel("Number of Orders")
plt.xlabel("Month")

#Step 6: Display the chart
plt.show()
```

```python
#PIE CHART
#Step 1 : Count each payment mode
payment_counts = df["Payment_Mode"].value_counts()

#Step 2: Create a pie chart (fixed repeated argument and typos)
plt.pie(payment_counts.values,
        labels=payment_counts.index,
        autopct="%1.1f%%")

#Step 3: Add title
plt.title("Payment Mode Share")

#Step 4: Display the chart
plt.show()
```

#Gap between Mean and Meridian

```python
# A big gap between mean and median is a warning sign
gap = pd.DataFrame({
    "Mean": raw[raw_cols].mean(),
    "Median": raw[raw_cols].median()
})

gap["Gap %"] = (((gap["Mean"] - gap["Median"]) / gap["Median"]) * 100).round(1)
gap.round(1)
```

```python
rules = {
    "Passengers": (0, 180), # aircraft has 180 seats
    "Ticket_Price": (1000, 20000), # domestic low-cost fare in INR
    "Delay_Minutes": (0, 600), # a delay cannot be negative
    "Flight_Duration_Min": (60, 240), # domestic sector
    "Customer_Rating": (1, 5) # rating scale 1 to 5
}

for col, (low, high) in rules.items():
    bad = raw[(raw[col] < low) | (raw[col] > high)]
    print(col, "-> impossible values:", len(bad))
    if len(bad) > 0:
        print(bad[["Date", "Flight_ID", col]].to_string(index=False))
        print()
```

#Box Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns

def show_boxplots(data, title):
    fig, axes = plt.subplots(3, 3, figsize=(13, 10))
    axes = axes.flatten()
    for i, col in enumerate(raw_cols):
        sns.boxplot(y=data[col], ax=axes[i], color="skyblue")
        axes[i].set_title(col)
        axes[i].set_ylabel("")
    fig.suptitle(title, fontsize=15)
    plt.tight_layout()
    plt.show()

show_boxplots(raw, "BEFORE cleaning: Boxplots of Numeric Features")
```

#Retaining, removing and capping outliers

```python
# Create a copy
df = raw.copy()

# Find ratings below 1 or above 5
bad_values = (df["Customer_Rating"] < 1) | (df["Customer_Rating"] > 5)

# Count unusual ratings (using .sum() properly)
print("Unusual ratings:", bad_values.sum())

# Replace unusual ratings with missing values
df.loc[bad_values, "Customer_Rating"] = np.nan

# Compare missing cells (fixed parentheses and syntax errors)
print("Missing cells before:", raw.isnull().sum().sum())
print("Missing cells after:", df.isnull().sum().sum())
```
